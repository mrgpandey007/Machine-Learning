```python
from sklearn.preprocessing import StandardScaler
cols=['Is_Businessclass','duration','airline_Vistara']
scaler= StandardScaler()
df[cols]=scaler.fit_transform(df[cols])
df.head()
```

```python
cat_category = {
   'Is_Businessclass','duration','days_left','airline_Vistara'}
```

```python
from scipy.stats import chi2_contingency
alpha=0.5
df['price_bin']=pd.qcut(df['price'],q=5,labels=False)
chi2={}
for col in cat_category:
    contin=pd.crosstab(df[col],df['price_bin'])
    chi2_stat,p_val,_,_=chi2_contingency(contin)
    decision = 'reject null' if p_val<alpha else 'Accepted Null'
    chi2[col]={
        'chi2_statistic':chi2_stat,
        'p_value':p_val,
        'Decision':decision
    }
chi2_d=pd.DataFrame(chi2).T
chi2_d=chi2_d.sort_values(by='p_value')
chi2_d
```
